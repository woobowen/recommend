```python
import tensorflow as tf
import gzip

file_path = "/data/mhwang/Rec/GRID/amazon_data/beauty/evaluation/partition_0.tfrecord.gz" # 假设文件已存在

# 使用 gzip.open 打开文件进行读取
with gzip.open(file_path, 'rb') as f:
    # tf.io.tf_record_iterator 会从文件句柄中读取记录
    # 注意：这种方式需要将整个解压后的文件读入内存，不适合大文件
    content = f.read()
    for record in tf.io.tf_record_iterator(iter([content])):
        # 手动解析
        example = tf.train.Example()
        example.ParseFromString(record)
        
        # 访问数据
        label = example.features.feature['label'].int64_list.value[0]
        print(f"读取到标签: {label}")
```
